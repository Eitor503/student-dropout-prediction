Welcome, the objective in this Notebook is preparing our current dataset with the information we collected on the past Notebook. So our df can be ready for training our model.

```python
import pandas as pd
```

```python
PATH = "../data/raw/dataset.csv"

df = pd.read_csv(PATH)
```

On the diagnosis, we saw that in the target value we had three different possible options: "Graduate", "Dropout" and "Enrolled".

That would make a mess in our model, so, instead of deleting those rows (that are around 18% of the whole Data Set), I decided to asign it the same numerical value as the "Graduate" target. To sumarize, would be:

Graduate: 0

Enrolled: 0

Dropout: 1

Students with an 'Enrolled' status were labeled as 0 (non-dropout) under the assumption that their current status reflects their final outcome. While a subset of these students may drop out in the future—introducing some label noise rather than data leakage—this approach establishes a pragmatic target variable for the predictive framework.

```python

# A dictionary with the changes
changes = {"Graduate": 0, "Enrolled": 0, "Dropout": 1}

NEW_PATH = "../data/processed/dataset.csv"

# And finally replace values
df["Target"] = df["Target"].replace(changes)

df.to_csv(NEW_PATH, index= False)
```
